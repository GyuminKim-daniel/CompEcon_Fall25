Exercise 15.1 : Using the above code blocks as an example, simulate the effects of a change in the corporate income tax rate (note the default in OG-Core is 21%). Just simulate the effects on the steady state (see the keyword arguments to the runner function). Warning: OG-Core allows the corporate income tax rate to vary over time and across industries, so you will need to enter it as a list of lists (even if it’s just a scalar that applies to all industries and time).

```python
"""
Exercise 15.1 — Steady-State Effects of a Corporate Income Tax Change (OG-Core)
-------------------------------------------------------------------------------

This script performs the following steps:

1. Solve the *baseline* OG-Core steady state (with the default CIT rate = 21%).
2. Solve a *reform* steady state that changes the corporate income tax rate.
   - In OG-Core, the CIT rate must be entered as a list of lists,
     since the model allows time- and industry-varying CIT rates.
     For a single industry and time-invariant tax:
         cit_rate = [[new_rate]]
3. Load the steady-state variable dictionaries from the output folders.
4. Compare baseline vs. reform GDP (Y).

Notes:
 - time_path=False ensures that OG-Core computes ONLY the steady state,
   not the full transition path.
 - OUTPUT_BASELINE and OUTPUT_REFORM will contain SS results and logs.
"""

from ogcore.parameters import Specifications
from ogcore.execute import runner
from pathlib import Path
import pickle
import numpy as np

# --------------------------------------------------------
# 1. Baseline model (default CIT = 21%)
# --------------------------------------------------------
p_base = Specifications(
    baseline=True,
    output_base="./OUTPUT_BASELINE"
)

print("Solving baseline steady state...")
baseline_results = runner(p_base, time_path=False)
print("Baseline steady state complete.\n")

# --------------------------------------------------------
# 2. Reform model: Change CIT rate to 15%
#    IMPORTANT: OG-Core requires a list of lists for cit_rate
# --------------------------------------------------------
p_reform = Specifications(
    baseline=False,
    output_base="./OUTPUT_REFORM"
)

p_reform.update_specifications({
    "cit_rate": [[0.15]]   # 15% CIT rate
})

print("Solving reform steady state (CIT=15%)...")
reform_results = runner(p_reform, time_path=False)
print("Reform steady state complete.\n")

# --------------------------------------------------------
# 3. Load steady-state variable pickle files
# --------------------------------------------------------
ss_base_path = Path("./OUTPUT_BASELINE/SS/SS_vars.pkl")
ss_reform_path = Path("./OUTPUT_REFORM/SS/SS_vars.pkl")

with ss_base_path.open("rb") as f:
    ss_base = pickle.load(f)

with ss_reform_path.open("rb") as f:
    ss_reform = pickle.load(f)

# --------------------------------------------------------
# 4. Compare GDP (Y) between baseline and reform
# --------------------------------------------------------
Y_base = ss_base["Y"]
Y_reform = ss_reform["Y"]
pct_change = 100 * (Y_reform - Y_base) / Y_base

print("Baseline GDP:", Y_base)
print("Reform GDP:", Y_reform)
print(f"GDP % Change from CIT cut: {pct_change:.2f}%")
```

```python
"""
Exercise 16.1 — Effect of Changing the Discount Factor β on Steady-State Savings
--------------------------------------------------------------------------------

This script solves two steady states in OG-Core:

1. Baseline:
   - Uses default household discount factor beta_annual.

2. Reform:
   - Changes the rate of time preference by modifying beta_annual.
   - A lower beta_annual makes households more impatient and reduces savings.
   - A higher beta_annual makes households more patient and increases savings.

After solving the two steady states, the script loads the saved SS variables
and uses ogcore.output_plots.ss_profiles to plot lifetime savings profiles
(b_{s+1,j}) for each ability type j.

IMPORTANT:
- OG-Core stores savings as **"b_sp1"** (the assets chosen at age s and held at s+1).
- ss_profiles() must be called with var="b_sp1" to show savings.
"""

import ogcore
from ogcore.parameters import Specifications
from ogcore import execute
from ogcore.output_plots import ss_profiles

import numpy as np
import pickle
from pathlib import Path
import copy

# ------------------------------------------------
# Working directories
# ------------------------------------------------
BASE = Path("C:/Users/Gyumin Kim/Desktop/Coursework/25-2R/Computational Methods for Economists/Problemset9")
BASE.mkdir(exist_ok=True)

OUT_BASELINE = BASE / "OUTPUT_BASELINE"
OUT_REFORM   = BASE / "OUTPUT_REFORM"

OUT_BASELINE.mkdir(exist_ok=True)
OUT_REFORM.mkdir(exist_ok=True)

# ------------------------------------------------
# 1. Baseline steady state
# ------------------------------------------------
p_base = Specifications(
    baseline=True,
    output_base=str(OUT_BASELINE),
    baseline_dir=str(OUT_BASELINE)
)

print("Running baseline steady state...")
execute.runner(
    p_base,
    time_path=False
)
print("Baseline SS complete.")

# ------------------------------------------------
# 2. Reform steady state — change β (beta_annual)
# ------------------------------------------------
p_reform = copy.deepcopy(p_base)

# Example: make households more impatient (β = 0.90)
p_reform.beta_annual = 0.90

# Reform flags
p_reform.baseline = False
p_reform.output_base = str(OUT_REFORM)
p_reform.baseline_dir = str(OUT_BASELINE)

print("Running reform steady state with new beta_annual...")
execute.runner(
    p_reform,
    time_path=False
)
print("Reform SS complete.")

# ------------------------------------------------
# 3. Load SS variables
# ------------------------------------------------
with open(OUT_BASELINE / "SS/SS_vars.pkl", "rb") as f:
    ss_vars_base = pickle.load(f)

with open(OUT_REFORM / "SS/SS_vars.pkl", "rb") as f:
    ss_vars_reform = pickle.load(f)

# ------------------------------------------------
# 4. Plot savings profiles using ss_profiles
# ------------------------------------------------
fig = ss_profiles(
    ss_vars_base, p_base,
    ss_vars_reform, p_reform,
    var="b_sp1",
    by_j=False
)

```

Exercise 16.2 Take TPI output from https://github.com/PSLmodels/OG-Core/blob/master/tests/test_io_data/TPI_vars_baseline.pkl and convert the GDP series to constant units of local currency using the factor_ss and the growth rates, g_n and g_y . You can find the factor_ss in the steady state output at https://github.com/PSLmodels/OG-Core/blob/master/tests/test_io_data/SS_vars_baseline.pkl and the growth rates in the parameters file at https://github.com/PSLmodels/OG-Core/blob/master/tests/test_io_data/model_params_baseline.pk Hint: you can look at the OG-Core chapter on stationarizing the model to help see how to get back to non-stationary measures.

```python
"""
Exercise 16.2 — Converting Stationary TPI GDP Back to Local Currency Units
--------------------------------------------------------------------------

This script loads three OG-Core test baseline files:

1. TPI_vars_baseline.pkl  – contains stationary transition-path variables (including Y_hat)
2. SS_vars_baseline.pkl   – contains the steady-state scaling factor "factor"
3. model_params_baseline.pkl – contains the parameters, including growth rates

OG-Core stationarizes aggregate variables using:

    Y_hat_t = Y_t / [ (1+g_y)^t * (1+g_n)^t * factor ]

To recover non-stationary GDP in local currency:

    Y_t = Y_hat_t * (1+g_y)^t * (1+g_n)^t * factor

Where:
 - g_y is the annual productivity growth rate
 - g_n is the steady-state population growth rate
 - factor is the scaling factor from the steady state

This script reconstructs Y_t and outputs a table containing:
 - stationary GDP
 - unstationarization growth factors
 - GDP in local currency units
"""

import pickle
import numpy as np
import requests
from io import BytesIO
import pandas as pd

# ---------------------------------------------------------
# Load baseline TPI variables
# ---------------------------------------------------------
url_tpi = "https://github.com/PSLmodels/OG-Core/raw/master/tests/test_io_data/TPI_vars_baseline.pkl"
tpi = pickle.load(BytesIO(requests.get(url_tpi).content))

# Load steady-state output (contains factor)
url_ss = "https://github.com/PSLmodels/OG-Core/raw/master/tests/test_io_data/SS_vars_baseline.pkl"
ss = pickle.load(BytesIO(requests.get(url_ss).content))

# Load parameters (Specifications object)
url_params = "https://github.com/PSLmodels/OG-Core/raw/master/tests/test_io_data/model_params_baseline.pkl"
params = pickle.load(BytesIO(requests.get(url_params).content))

# ---------------------------------------------------------
# Extract needed objects
# ---------------------------------------------------------
Y_stationary = tpi["Y"]      # stationary GDP series
factor_ss = ss["factor"]     # scaling factor from SS

# Growth rates
g_y = params.g_y_annual           # annual productivity growth
g_n = params.g_n_ss               # steady-state population growth

T = len(Y_stationary)

# ---------------------------------------------------------
# Undo OG-Core stationarization
# ---------------------------------------------------------

# Productivity growth term
G_prod = (1 + g_y) ** np.arange(T)

# Population growth term
G_pop = (1 + g_n) ** np.arange(T)

# Convert stationary → local currency GDP
Y_local = Y_stationary * G_pop * G_prod * factor_ss

# ---------------------------------------------------------
# Results table
# ---------------------------------------------------------
df = pd.DataFrame({
    "Y_stationary": Y_stationary,
    "G_pop": G_pop,
    "G_prod": G_prod,
    "Y_local_currency": Y_local
})

df.head(10)
```

```python
"""
Exercise 16.5 — Baseline vs Reform GDP Forecast in Levels (OG-Core)
------------------------------------------------------------------

This script uses OG-Core's plot_aggregates function to compare how a
baseline GDP forecast would differ under the reform scenario.

Steps:
1. Load baseline and reform TPI variables and model parameters.
2. Construct a baseline GDP forecast in **levels**, using an assumed growth
   rate. (Here we use a simple 2% annual growth forecast, but any external
   forecast could be inserted.)
3. Call ogcore.output_plots.plot_aggregates with:
       - plot_type="forecast"
       - forecast_data=<GDP forecast array>
4. OG-Core then adjusts this forecast using the differences between
   the reform and baseline transition paths for GDP, producing a
   reform-adjusted forecast.

Important notes:
 - forecast_data must have the same length as num_years_to_plot.
 - var_list=["Y"] ensures that only GDP is plotted.
 - The forecast is in **levels**, not percent differences.
"""

import pickle
import numpy as np
import matplotlib.pyplot as plt
from ogcore.output_plots import plot_aggregates

# ---------------------------------------------------------
# Load baseline and reform TPI + parameter files
# ---------------------------------------------------------
with open("TPI_vars_baseline.pkl", "rb") as f:
    TPI_baseline = pickle.load(f)

with open("model_params_baseline.pkl", "rb") as f:
    params_baseline = pickle.load(f)

with open("TPI_vars_reform.pkl", "rb") as f:
    TPI_reform = pickle.load(f)

with open("model_params_reform.pkl", "rb") as f:
    params_reform = pickle.load(f)

# ---------------------------------------------------------
# Create a baseline GDP forecast (LEVELS)
# ---------------------------------------------------------
num_years = 50
initial_GDP = TPI_baseline["Y"][0]

# Example assumption: 2% annual GDP growth
g_forecast = 0.02

baseline_GDP_forecast = initial_GDP * (1 + g_forecast) ** np.arange(num_years)

# ---------------------------------------------------------
# Plot: Baseline GDP Forecast vs Reform-Adjusted Forecast
# ---------------------------------------------------------
fig = plot_aggregates(
    TPI_baseline,
    params_baseline,
    TPI_reform,
    params_reform,
    var_list=["Y"],                    # Only GDP
    plot_type="forecast",              # Required for Exercise 16.5
    forecast_data=baseline_GDP_forecast,
    forecast_units="Levels",
    num_years_to_plot=num_years
)

plt.show()
```

Exercies 17.2 : Using your modified demographics.py, plot the fertility rates in this country. Note that you can do this directly from the demographics.get_fert() function.

```python
"""
Exercise 17.2 — Plot Fertility Rates Using demographics.get_fert()
------------------------------------------------------------------

We first retrieve raw UN fertility data (variable 68) to determine the
actual age grid (the UN WPP dataset often uses 5-year age groups,
e.g. ages 15,20,25,...,45). We then call get_fert() with matching
min_age, max_age, and totpers so that pop_rebin() does not crash.

This is the intended workflow in the reading: use your modified
demographics.py together with get_fert().
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ogcore.demographics import get_un_data, get_fert

# -------------------------------------------
# Step 1: Get raw fertility data (UN variable 68)
# -------------------------------------------
country_id = "826"   # UK
start_year = 2020
end_year   = 2050

fert_raw = get_un_data(
    "68",
    country_id=country_id,
    start_year=start_year,
    end_year=end_year
)

print("Raw fertility data:")
print(fert_raw.head())

# -------------------------------------------
# Step 2: Determine correct fertility age bins
# -------------------------------------------
ages = sorted(fert_raw["age"].unique())
min_age = min(ages)
max_age = max(ages)
totpers = len(ages)

print("min_age:", min_age)
print("max_age:", max_age)
print("totpers:", totpers)

# -------------------------------------------
# Step 3: Call get_fert() using correct age grid
# -------------------------------------------
fert_rates = get_fert(
    totpers=totpers,
    min_age=min_age,
    max_age=max_age,
    country_id=country_id,
    start_year=start_year,
    end_year=end_year,
    graph=False
)

# fert_rates shape = (#years , #age bins)
years = list(range(start_year, end_year+1))

fert_df = pd.DataFrame(fert_rates, index=years, columns=ages)

# -------------------------------------------
# Plot fertility for first and last year
# -------------------------------------------
fert_first = fert_df.loc[start_year]
fert_last  = fert_df.loc[end_year]

plt.figure(figsize=(10,6))
plt.plot(ages, fert_first, label=str(start_year))
plt.plot(ages, fert_last,  label=str(end_year))

plt.title(f"UK Fertility Rates by Age (using get_fert)")
plt.xlabel("Age of Mother")
plt.ylabel("Fertility Rate (births per woman)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
"""
Exercise 17.6 — Effect of Alternative Demographics on OG-Core Steady State
--------------------------------------------------------------------------

This exercise:
 1. Runs a baseline OG-Core steady state using default demographics.
 2. Constructs new population objects (omega, g_n, mortality, fertility, immigration)
    using get_pop_objs().
 3. Inserts these demographics into a reform Specifications object.
 4. Solves the new steady state.
 5. Compares baseline vs reform steady-state macro variables using macro_table_SS.
"""

import os
from pathlib import Path
import pickle
from ogcore.parameters import Specifications
from ogcore.execute import runner
from ogcore.demographics import get_pop_objs
from ogcore.output_tables import macro_table_SS

# ---------------------------------------------------------
# 1) Create output directories for baseline and reform runs
# ---------------------------------------------------------
BASE_DIR = Path("OGCORE_OUTPUT_BASELINE")
REFORM_DIR = Path("OGCORE_OUTPUT_REFORM")

BASE_DIR.mkdir(parents=True, exist_ok=True)
REFORM_DIR.mkdir(parents=True, exist_ok=True)

# ---------------------------------------------------------
# 2) Baseline steady-state simulation
# ---------------------------------------------------------
p_base = Specifications(
    baseline=True,
    baseline_dir=str(BASE_DIR),
    output_base=str(BASE_DIR),
)

print("Running BASELINE steady state...")
runner(p_base)
print("Baseline completed.")

# ---------------------------------------------------------
# 3) Construct population objects with updated demographics
# ---------------------------------------------------------
E = 20
S = 80
T = 320
curr_year = 2023   # as in exercise prompt

pop_objs = get_pop_objs(E=E, S=S, T=T, initial_data_year=curr_year)

# ---------------------------------------------------------
# 4) Reform steady-state simulation using updated demographics
# ---------------------------------------------------------
p_reform = Specifications(
    baseline=False,
    baseline_dir=str(BASE_DIR),
    output_base=str(REFORM_DIR),
)

# Insert population objects into reform specifications
p_reform.update_specifications(pop_objs)

print("Running REFORM steady state with updated demographics...")
runner(p_reform)
print("Reform completed.")

# ---------------------------------------------------------
# 5) Load steady-state outputs
# ---------------------------------------------------------
def load_pkl(path):
    with open(path, "rb") as f:
        return pickle.load(f)

ss_base = load_pkl(BASE_DIR / "SS/SS_vars.pkl")
params_base = load_pkl(BASE_DIR / "model_params.pkl")

ss_reform = load_pkl(REFORM_DIR / "SS/SS_vars.pkl")
params_reform = load_pkl(REFORM_DIR / "model_params.pkl")

# ---------------------------------------------------------
# 6) Compare steady-state macro outcomes
# ---------------------------------------------------------
table_ss = macro_table_SS(
    ss_base,
    params_base,
    ss_reform,
    params_reform,
    output_type=None
)

print("\n=== STEADY STATE COMPARISON TABLE ===")
print(table_ss)
```

Exercise 17.10 : Now let’s estimate the autocorrelation in GDP growth, assuming is a first order autoregressive process (AR1). We’ll do this with the statsmodles.OLS function. First, create a new column in your data frame that is the lagged value of GDP growth. Then, use the statsmodels.regression.linear_model.OLS function to estimate the following equation: What did you estimate as ? What does this represent? How persistent is the growth process in South Africa?

```python
"""
Exercise 17.10
Estimate AR(1) autocorrelation in South Africa's annual GDP growth using:

    statsmodels.regression.linear_model.OLS

Steps:
 1. Download quarterly real GDP (SA).
 2. Convert to annual GDP.
 3. Compute annual growth rate.
 4. Create lagged growth column.
 5. Estimate AR(1): g_t = α + φ g_{t-1} + ε_t
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pandas_datareader.data import DataReader
from statsmodels.regression.linear_model import OLS
import statsmodels.api as sm

# ---------------------------------------
# 1. Download quarterly real GDP (SA)
# ---------------------------------------
fred_series = "NGDPRNSAXDCZAQ"
gdp_q = DataReader(fred_series, "fred", start="1990-01-01").dropna()

# ---------------------------------------
# 2. Convert to annual GDP (sum of 4 quarters)
# ---------------------------------------
gdp_annual = gdp_q.resample("A").sum()
gdp_annual.index = gdp_annual.index.year   # convert to year integers

# ---------------------------------------
# 3. Compute annual GDP growth (%)
# ---------------------------------------
gdp_growth = gdp_annual.pct_change() * 100
gdp_growth = gdp_growth.dropna()

# ---------------------------------------
# 4. Prepare AR(1) regression dataset
# ---------------------------------------
df = pd.DataFrame({"g": gdp_growth[fred_series]})
df["g_lag"] = df["g"].shift(1)
df = df.dropna()

Y = df["g"]
X = sm.add_constant(df["g_lag"])  # add intercept

# ---------------------------------------
# 5. Estimate AR(1) using OLS
# ---------------------------------------
model = OLS(Y, X).fit()

phi_hat = model.params["g_lag"]

print("Estimated AR(1) φ =", round(phi_hat, 4))
print("\n=== Regression Summary ===")
print(model.summary())

# ---------------------------------------
# 6. Plot growth series + lag for visualization
# ---------------------------------------
plt.figure(figsize=(10,5))
plt.plot(df.index, df["g"], label="GDP Growth (%)")
plt.plot(df.index, df["g_lag"], label="Lagged GDP Growth (%)", alpha=0.6)
plt.legend()
plt.grid(True)
plt.title("South Africa Annual GDP Growth and Lagged Growth")
plt.xlabel("Year")
plt.ylabel("Growth (%)")
plt.show()
```

(2) What does the constant represent?
The constant alpha represetns the baseline or trend component of GDP growth after accouting for last year's growth. It reflects a sample-specific correction term in a dataset that behaves almost like a random walk with drift.

Exercise 17.12 : Now we’re going to adjust the above earnings process to match the Gini coefficient in a country that isn’t the U.S. (note that the Gini coefficient for income in the U.S. was 41.5 in 2019). The transformation we’ll make is: where represents the country of interest. We will find by matching the Gini coefficient in the country of interest. To do this, we’ll use the scipy.optimize module. First, create a function that takes as an  argument and returns the Gini coefficient in the country of interest (hint: the ogusa.utils.Inequality class object will be helpful here). Then, use the scipy.optimize.minimize_scalar function to find the value of that minimizes the difference between the Gini coefficient in the country of interest and the Gini coefficient in the U.S. Finally, use the value of you found to transform the e matrix from the U.S. to the country of interest. Plot the earnings processes in the U.S. and the country of interest. How do they differ?

```python
"""
Exercise 17.12 — Adjust earnings process to match a country's Gini
------------------------------------------------------------------
We adjust the U.S. earnings matrix e(s,j) by applying a nonlinear
transformation e_country = e_US ** theta so that the earnings-induced
Gini coefficient matches a target country’s Gini.
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar
from income import get_e_orig   # OG-USA earnings process
from ogusa_utils import Inequality

# ------------------------------------------------------------
# Step 1: Load U.S. earnings matrix e(s,j) from income.py
# ------------------------------------------------------------
S = 80
J = 7

age_wgts = np.ones(S) / S
abil_wgts = np.array([0.25, 0.25, 0.20, 0.10, 0.10, 0.09, 0.01])

# Load OG-USA original earnings matrix
e_US = get_e_orig(age_wgts, abil_wgts)   # shape (80,7)
print("Earnings matrix shape:", e_US.shape)

# ------------------------------------------------------------
# Step 2: Function to compute Gini using OG-USA Inequality class
# ------------------------------------------------------------
def compute_gini(e_matrix):
    ineq = Inequality(
        dist=e_matrix,
        pop_weights=age_wgts,
        ability_weights=abil_wgts,
        S=S,
        J=J
    )
    return float(ineq.gini())

print("U.S. earnings Gini:", compute_gini(e_US))

# ------------------------------------------------------------
# Step 3: Define Gini(theta) for transformed earnings
# ------------------------------------------------------------
def gini_of_theta(theta):
    theta = float(theta)
    return compute_gini(e_US ** theta)

# ------------------------------------------------------------
# Step 4: Define target country Gini and objective function
# ------------------------------------------------------------
gini_target = 0.45     # Example target country

def objective(theta):
    return (gini_of_theta(theta) - gini_target)**2

# ------------------------------------------------------------
# Step 5: Solve for theta that hits target Gini
# ------------------------------------------------------------
sol = minimize_scalar(objective, bounds=(0.2, 3.0), method="bounded")
theta_hat = sol.x

print("\n====== Optimization Results ======")
print("Converged:", sol.success)
print(f"theta_hat = {theta_hat:.4f}")
print(f"Gini(theta_hat) = {gini_of_theta(theta_hat):.4f}")
print("Target country Gini:", gini_target)

# ------------------------------------------------------------
# Step 6: Construct transformed earnings matrix
# ------------------------------------------------------------
e_country = e_US ** theta_hat

# ------------------------------------------------------------
# Step 7: Plot earnings profiles
# ------------------------------------------------------------
ages = np.linspace(21, 100, S)

plt.figure(figsize=(10,6))
for j in range(J):
    plt.plot(ages, e_US[:, j], label="U.S." if j == 0 else "", color="blue", alpha=0.5)
    plt.plot(ages, e_country[:, j], "--", label="Adjusted" if j == 0 else "", color="red", alpha=0.7)

plt.title(f"Earnings Profiles: U.S. vs. Adjusted (theta = {theta_hat:.3f})")
plt.xlabel("Age")
plt.ylabel("Earnings Ability")
plt.grid(True)
plt.legend()
plt.show()
```

The transformed earnings profiles differ from the U.S. profiles because the estimated θ = 1.155 is greater than one. This amplifies earnings for higher-ability types much more than for lower-ability types. As a result, the vertical distance between ability groups becomes larger, especially near peak working ages. The age-earnings shape (the hump around ages 50–60) remains essentially the same, but it is scaled upward for high-ability households.
